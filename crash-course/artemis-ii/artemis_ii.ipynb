### 1.2 Types

Python has many built-in variable **types**. The type of a variable can be found with the function `type()`. Four simple variable types are as follows:

- `int`: An integer number. Unlike many other coding languages, an integer in Python can be of any length. Integers are initialized as whole numbers (e.g., `ex_variable = 1`).
- `float`: A decimal number. For very large or very small floats, they're often abbreviated into the form `XeY`, equivalent to $X \times 10^Y$ for numbers `X` and `Y`. Floats are initialized as decimal-valued numbers (e.g., `ex_variable = 1.05`).
- `bool`: A boolean, or a true/false value. Booleans are initialized as `True` or `False` (e.g., `ex_variable = False`).
- `str`: A string, or a sequence of text characters. Strings are initialized as a sequence of characters surrounded by quotation marks (e.g., `ex_variable = "False"` or `ex_variable = 'Elephant'`).

This explains why `print("Hello World!")` above simply output `Hello World!` &mdash; the quotation marks indicated that the characters contained within were to be read as a string-type variable, and thus the value of that string was output. 

**Example:**

```python
# Define a string and a boolean value
name_of_cat = "Jonathan"
is_cat = True
print(type(is_cat)) # Should return <class 'bool'>
```

### 1.3 Operations

Python also has built-in support for simple **operations** between variables. These include mathematical operations (applicable to both integers and floats) and boolean operations (applicable to all types). To apply an operation between variables, simply write `a (operator) b`, where `a` and `b` are the variables or variable names in question.

Python supports the four classical mathematical operations: `+`, `-`, `*`, and `/`. Mathematical operations also include the modulus operator `%` (where `a % b` outputs the remainder of the division operation $a/b$), and the exponentiation operator `**` (remark: the `^` operator also exists, but is *not* an exponentiation operator).

As with mathematics, parentheses indicate to Python the order in which to evaluate operations.

**Example:**

```python
# Define two integers and output their product
number_of_cities = 64
people_per_city = 1_000_000 # You can include underscores in integers where commas would normally go, to make large integers easier to read
total_population = number_of_cities * people_per_city
print(total_population)
```

The boolean operators `>`, `<`, `>=`, and `<=` also exist, but they work only on variables whose values can be compared (typically just numerical variables).

## **2. Lists**

### 2.1 Definition

A `list` is another fundamental way to store information in Python. As the name suggests, they are useful for organizing many individual pieces of information into one place for later reference. 

Python lists are rather unique: they are *mutable* (can be changed after being created), contain an *ordered* sequence of elements, and the elements can be any data type (integers, strings, floats, etc.) or a mixture of data types. 

Lists can be defined just ike any other variable with the single equal sign operator `=`. To fill a list with **elements** at its definition call, we use square brackets `[ ]` and commas `,` to separate each element. Let's now make a list.

**Exercise**

```python
# Defining a list variable
# Have the class decide to add a random string to our list!
list1 = [5, 8, "PLACEHOLDER", -22, True, 0.1, "hello", 8.25] #### Remove PLACEHOLDER
```

As mentioned above, one unique thing about Python lists is that they let you hold different data types in the same list! Lists allow us to store as many variables as we want, even other lists (this is called a *nested list*). 

Of course, the real usefulness of list is the user's ability to **index** them. This means that we can access any element in the list whenever we want and wherever it is in the list, without altering the list itself. 

To access an element from a list, we can call the name of the list variable followed by square brackets `[ ]` containing that element's **index number**.

**Example:**

```python
# Get an item from the list
item1 = list1[2]
print(item1)
```

### 2.2 Slicing Lists

Note that while we asked for element index 2 in the example above, it gave us the 3rd item in the list. This is normal: Python starts indexing elements with the number 0, for historical reasons. *Keep in mind, this implies that the last element in a list always has an index equal to the length of the list minus one!*

We can also get multiple values at once from a list by **slicing** the list using the colon symbol `:` and two index values. Note that the ending index is not included in the slice, while the starting index *is*.

**Example:**

```python
# Get a slice of a list
print(list1[3:6])
```

### 2.3 Appending and Removing Elements of a List
The most important operation to learn here is how to add an element to an already-existing `list`. Elements can be added using the `.append()` method, which is written immediately following the variable name of a given list.

**Example:**

```python
# Make a new list
list2 = [1, 2, 4, 8]
print(list2)

# Add to the list
list2.append("Sixteen")
print(list2)
```

### 2.4 Nested Lists

One of the strengths of using a `list` in Python is that they can contain any variables of any type — even lists! This means that you can have nested lists of data which is enormously helpful for saving multiple columns of data from a file. Let's define a list that contains our list1 and list2 variables, in addition to some other elements:

**Example:**

```python
# Example of nested list:
list3 = [list1, "apple", 3.14159, list2, "pie"]

print('Length of list3:', len(list3))
print('My nested list3:', list3)
```

We can see that our `len()` function returned the number of elements or rows in our nested list, which represent the first layer of data. To access each element or row in this 2-layered nested list, we can therefore use the same indexing notation as before:

**Example:**

```python
print("Each element in list3:")
print(list3[0])
print(list3[1])
print(list3[2])
print(list3[3])
print(list3[4])
```

Nested lists and indexing is an incredibly powerful tool for data analysis, as most data formats you will work with have at least two layers (rows and columns).

### 2.5 Application to Artemis II

Now that we have a better understanding of lists, we can begin to work with the Artemis II trajectory data that we previously imported. As a reminder, this data looks as follows:

The `trajectory` variable contains every recorded position of the Artemis II spacecraft, where each row holds two values: the **x-cordinate** (column 0) and the **y-coordinate** (column 1) of the spacecraft, measured in kilometers from Earth's center. 

Let's begin by exploring the structure of our data:

**Example:**

```python
# What do the first few rows look like? 
print("First 3 positions:\n", trajectory[0:3])
```

##### Utilizing Artemis II Data 

Each element of `trajectory` is itself a two-element list, just like the nested lists we saw above! We can therefore use **double-indexing** to pull out a single coordinate: `trajectory[row][column]`, where column `0` is x and column `1` is y.

<center><img src="https://raw.githubusercontent.com/pyiu-workshops/pyiu/refs/heads/develop/crash-course/artemis-ii/imgs/artemis_ii_double_indexing.jpg" width="600" alt="Annotated double indexing infographic"></center>

**Exercise:** Let's practice by separating the x- and y-coordinates of the first three data points into their own lists.

```python
# Separate x- and y-coordinates for the first 3 data points
x_coords = []
x_coords.append(trajectory[0][0])  # x-coordinate of Artemis II position 0
x_coords.append(trajectory[1][0])  # x-coordinate of Artemis II position 1 #### Remove both indices
x_coords.append(trajectory[2][0])  # x-coordinate of Artemis II position 2 #### Remove both indices

y_coords = []
y_coords.append(trajectory[0][1])  # y-coordinate of Artemis II position 0
y_coords.append(trajectory[1][1])  #### Remove both indices
y_coords.append(trajectory[2][1])  #### Remove both indices

print("x-coordinates (km):", x_coords)
print("y-coordinates (km):", y_coords)
```
